This folder is a compact reading edition of the analysis workflows.
Open a tutorial below. The notebooks contain workflow text, code and saved outputs. Compartment retains the original upstream notebook; reading a preview does not run an analysis.

## 3. Read the tutorials and inspect their implementation

This public test repository contains seven reading notebooks, six task webpages and their web index. The notebooks retain existing figures; Compartment preserves the original code cells and saved outputs.

Read the [published website](https://rawandy2001-design.github.io/benchmark-tutorials/) or open the notebooks in Jupyter. The [companion analysis repository](https://github.com/rawandy2001-design/benchmark) contains the implementation scripts, shared path setup and task-specific environment instructions; access to that private repository is required to open its source links.

Command examples refer to that analysis-code checkout. Running them requires the corresponding inputs, checkpoints and native environments.

The presentation follows the official scHiCluster [imputation](https://zhoujt1994.github.io/scHiCluster/Tan2021/imputation.html) and [loop-analysis](https://zhoujt1994.github.io/scHiCluster/hba/loop_majortype/01.diff.html) tutorial style.

### Shared path prefix

All tutorial commands use `BENCHMARK_ROOT`. The local workspace contains
directory links to the existing code, inputs, software and saved records.
The source files remain in their original locations.

Run this once from the code checkout in WSL:

```bash
source tutorials/paths.sh
cd "${BENCHMARK_ROOT}/code"
```

The local prefix is `/mnt/f/project/test1/benchmark_workspace` in WSL and
`F:/project/test1/benchmark_workspace` in Windows. For a Windows Python session,
set the corresponding environment variable in PowerShell:

```powershell
$env:BENCHMARK_ROOT = "F:/project/test1/benchmark_workspace"
```

Python examples use:

```python
from pathlib import Path
import os

BENCHMARK_ROOT = Path(os.environ["BENCHMARK_ROOT"])
```

```text
BENCHMARK_ROOT/
  code/                    analysis code checkout
  data/higashi/            Lee2019 Embedding and TAD inputs
  data/lee2019/            Lee2019 pseudo-bulk matrices and cell tables
  data/mousebrain/         matched mouse-brain contact maps and loops
  data/rna/                expression and DEG inputs
  data/mcg/                Lee2019 MCDS
  records/                 saved Loop tables and run records
  software/HiCFoundation/  HiCFoundation source and checkpoints
  software/Mustache/       native Mustache environment
  output/                  outputs selected for new commands
```

The mappings and runtime settings are collected in [paths.sh](https://github.com/rawandy2001-design/benchmark/blob/main/tutorials/paths.sh).
On another machine, set its `BENCHMARK_*_SOURCE` values to the corresponding
existing directories. Original file excerpts retain the paths recorded in
their source files.
